```python
print(services.columns.tolist())
print(services.head())
print(staff.columns.tolist())
print(schedule.columns.tolist())
```

```python
patients["week"] = patients["arrival_date"].dt.isocalendar().week.astype(int)
```

```python
patient_summary = patients.groupby(
    ["week", "service"]
).agg(
    patient_count=("patient_id", "count"),
    average_age=("age", "mean"),
    average_length_of_stay=("length_of_stay", "mean"),
    average_satisfaction=("satisfaction", "mean")
).reset_index()
```

```python
print(patient_summary.shape)
print(patient_summary.head())
```

```python
staff_summary = staff.groupby(
    "service"
).agg(
    staff_count=("staff_id", "count")
).reset_index()

print(staff_summary)
```

```python
staff_weekly = schedule.groupby(
    ["week", "service"]
).agg(
    staff_present=("present", "sum")
).reset_index()

print(staff_weekly.head())
```

```python
hospital_raw = services.merge(
    patient_summary,
    on=["week", "service"],
    how="left"
)

print(hospital_raw.shape)
print(hospital_raw.head())
```

```python
hospital_raw = hospital_raw.merge(
    staff_summary,
    on="service",
    how="left"
)

print(hospital_raw.shape)
```

```python
hospital_raw = hospital_raw.merge(
    staff_weekly,
    on=["week", "service"],
    how="left"
)

print(hospital_raw.shape)
print(hospital_raw.head())
```

```python
print("Duplicate staff-week records:")
print(schedule.duplicated(["week", "staff_id"]).sum())

print("Total staff:", staff["staff_id"].nunique())
print("Scheduled staff:", schedule["staff_id"].nunique())

print("Staff IDs in schedule but not staff:")
print((~schedule["staff_id"].isin(staff["staff_id"])).sum())
```

```python
staff_check = schedule.merge(
    staff,
    on=["staff_name", "role", "service"],
    how="left",
    indicator=True
)

print(staff_check["_merge"].value_counts())
```

```python
unmatched = staff_check[
    staff_check["_merge"] == "left_only"
]

print(unmatched[
    ["staff_id_x", "staff_name", "role", "service", "week"]
].head(10))
```

```python
print(staff.head())
```

```python
schedule_names = set(schedule["staff_name"])
staff_names = set(staff["staff_name"])

print("Names only in schedule:", len(schedule_names - staff_names))
print("Names only in staff:", len(staff_names - schedule_names))
```

```python
staff_check = schedule.merge(
    staff,
    on="staff_name",
    how="left",
    suffixes=("_schedule", "_master"),
    indicator=True
)

print(staff_check["_merge"].value_counts())

matched = staff_check[staff_check["_merge"] == "both"]

print("Role mismatches:",
      (matched["role_schedule"] != matched["role_master"]).sum())

print("Service mismatches:",
      (matched["service_schedule"] != matched["service_master"]).sum())
```

```python
staff_summary = staff.groupby("service").agg(
    staff_count=("staff_id", "nunique")
).reset_index()

staff_weekly = schedule.groupby(
    ["week", "service"]
).agg(
    staff_present=("present", "sum")
).reset_index()

hospital_raw = services.merge(
    patient_summary,
    on=["week", "service"],
    how="left"
)

hospital_raw = hospital_raw.merge(
    staff_weekly,
    on=["week", "service"],
    how="left"
)

hospital_raw = hospital_raw.merge(
    staff_summary,
    on="service",
    how="left"
)

print("Integrated shape:", hospital_raw.shape)
print(hospital_raw.head())
```

```python
print("Missing values:")
print(hospital_raw.isnull().sum())

print("\nDuplicate rows:")
print(hospital_raw.duplicated().sum())

print("\nDataset shape:")
print(hospital_raw.shape)
```
