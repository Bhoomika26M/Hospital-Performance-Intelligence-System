```python
admissions = pd.read_csv(
    r"C:\Users\HP\OneDrive\Desktop\Hospital_Analytics_project\Hospital_admission dataset\hospital_admissions_cleaned.csv"
)
print(admissions.shape)
print(admissions.columns.tolist())
```

```python
readmission_rate = admissions["is_readmission_30d"].mean() * 100

print("Readmission Rate:", round(readmission_rate, 2), "%")
```

```python
print(df[[
    "week",
    "service",
    "available_beds",
    "patients_admitted",
    "patients_refused"
]].head(10))
```

```python
df["occupancy_rate"] = (
    df["patients_admitted"] / df["available_beds"]
) * 100

df["bed_utilization_rate"] = (
    df["patients_admitted"] * df["average_length_of_stay"]
) / (df["available_beds"] * 7) * 100

print(df[[
    "service",
    "occupancy_rate",
    "bed_utilization_rate"
]].head())
```

```python
df = df.drop(columns=["occupancy_rate", "bed_utilization_rate"])
```

```python
total_admissions = df["patients_admitted"].sum()

average_los = admissions["length_of_stay_days"].mean()

print("Total Admissions:", total_admissions)
print("Average Length of Stay:", round(average_los, 2), "days")
```

```python
readmission_rate = (admissions["is_readmission_30d"].mean() * 100)

print("Readmission Rate:", round(readmission_rate, 2), "%")
```

```python
df["admission_success"] = (
    df["patients_admitted"] / df["patients_request"]
).clip(0, 1) * 100

df["staff_coverage"] = (
    df["staff_present"] / df["staff_count"]
).clip(0, 1) * 100

df["satisfaction_score"] = (
    df["average_satisfaction"].clip(0, 100)
)

df["department_efficiency_score"] = (
    df["admission_success"] * 0.40
    + df["staff_coverage"] * 0.30
    + df["satisfaction_score"] * 0.30
)
print(df[[
    "service",
    "admission_success",
    "staff_coverage",
    "satisfaction_score",
    "department_efficiency_score"
]].head())
```

```python
print(df[[
    "service",
    "available_beds",
    "patient_count",
    "average_length_of_stay"
]].head(10))
```

```python
df["estimated_bed_utilization"] = (df["patient_count"] * df["average_length_of_stay"]/ (df["available_beds"] * 7)) * 100
print(df[[
    "service","estimated_bed_utilization"
]].head(10))
```

```python
df["total_admissions"] = total_admissions
df["overall_average_los"] = average_los
df["readmission_rate"] = readmission_rate

df.to_excel("hospital_final_dataset.xlsx", index=False)
```
